```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

from src.document_loader import load_all_pdfs

documents = load_all_pdfs(PROJECT_ROOT / "data" / "documents")

print("===== DOCUMENT COLLECTION OVERVIEW =====")
print("Total pages loaded:", len(documents))
print("Total source documents:", len(set(doc["source"] for doc in documents)))

print("\nDocuments:")
for source in sorted(set(doc["source"] for doc in documents)):
    pages = [doc["page"] for doc in documents if doc["source"] == source]
    print(f"- {source}: {len(pages)} pages")
```

```python
import pandas as pd

document_summary = []

for source in sorted(set(doc["source"] for doc in documents)):
    source_documents = [
        doc for doc in documents
        if doc["source"] == source
    ]

    document_summary.append({
        "Document": source,
        "Pages": len(source_documents),
        "First Page": min(doc["page"] for doc in source_documents),
        "Last Page": max(doc["page"] for doc in source_documents)
    })

document_df = pd.DataFrame(document_summary)

document_df
```
